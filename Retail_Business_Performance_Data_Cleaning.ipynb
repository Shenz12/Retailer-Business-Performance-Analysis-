**Import libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load dataset**

```python
df = pd.read_csv("Retail_Business_Performance_Dataset.csv")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.head()
```

**Check data types**

```python
df.info()
```

**Check missing values**

```python
df.isnull().sum()
```

**Check duplicates**

```python
print("Duplicate rows:", df.duplicated().sum())
```

**Convert date**

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
```

**Check duplicate Order IDs**

```python
print("Duplicate Order IDs:", df['Order_ID'].duplicated().sum())
```

**Check numerical columns**

```python
numeric_columns = [
    'Quantity',
    'Unit_Price',
    'Discount',
    'Sales',
    'Cost',
    'Profit',
    'Inventory_Days'
]

df[numeric_columns].describe()
```

## **Create important analytical columns**

**Profit Margin**

```python
df['Profit_Margin'] = (df['Profit'] / df['Sales']) * 100
```

**Revenue before discount**

```python
df['Gross_Sales'] = df['Quantity'] * df['Unit_Price']
```

**Discount Amount**

```python
df['Discount_Amount'] = df['Gross_Sales'] - df['Sales']
```

**Year**

```python
df['Year'] = df['Order_Date'].dt.year
```

**Month**

```python
df['Month'] = df['Order_Date'].dt.month
```

**Month Name**

```python
df['Month_Name'] = df['Order_Date'].dt.month_name()
```

**Save the cleaned dataset**

```python
output_file = "Retail_Business_Performance_Cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

print(
    f"Cleaned dataset saved as: {output_file}"
)
```
